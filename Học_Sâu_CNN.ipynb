```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sonalshinde123/tiger-vs-lion-image-classification-dataset")

print("Path to dataset files:", path)
```

```python
import os
os.listdir(path)
```

```python
os.listdir(os.path.join(path,'lion_tiger_dataset'))
```

```python
os.listdir(os.path.join(path,'lion_tiger_dataset/tiger'))
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.3
)
```

```python
train = datagen.flow_from_directory(
    os.path.join(path,'lion_tiger_dataset'),
    target_size=(224,224),
    batch_size=32,
    class_mode='binary',
    subset='training'
)
```

```python
val = datagen.flow_from_directory(
    os.path.join(path,'lion_tiger_dataset'),
    target_size=(224,224),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)
```

```python
print(train.class_indices)
```

```python
from keras.models import Sequential
from keras.layers import Conv2D,MaxPooling2D,Dense,Flatten,Dropout,BatchNormalization,GlobalAveragePooling2D
```

```python
import tensorflow as tf
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
])
```

```python


model = Sequential()

# Block 1
model.add(Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(224,224,3)))
model.add(Conv2D(32, (3,3), activation='relu', padding='same'))
model.add(MaxPooling2D())

# Block 2
model.add(Conv2D(64, (3,3), activation='relu', padding='same'))
model.add(Conv2D(64, (3,3), activation='relu', padding='same'))
model.add(MaxPooling2D())

# Block 3
model.add(Conv2D(128, (3,3), activation='relu', padding='same'))
model.add(Conv2D(128, (3,3), activation='relu', padding='same'))
model.add(MaxPooling2D())

# Classification head
model.add(GlobalAveragePooling2D())
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
history = model.fit(
    train,
    validation_data=val,
    epochs=5
)
```
